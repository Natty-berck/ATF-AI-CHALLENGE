### 1.2 Few-Shot Prompting

Unlike Zero-Shot prompting where we ask the model to perform a task with no examples, **Few-Shot Prompting** provides a few input-output examples to guide the model's behavior.

This is particularly powerful for:
- Controlling the exact output format (e.g., JSON, structured text).
- Guiding the style, tone, or specific behavior of your digital twin.
- Standardizing complex classifications.

````python
# Example of Few-Shot Prompting structure for Digital Twin style transfer
few_shot_prompt_template = """
You are a helpful assistant acting as my digital twin. Mirror my speaking style and tone based on these examples:

Input: How was your day?
Response: Honestly, it was pretty exhausting. Non-stop meetings, but we got a lot done!

Input: Do you want to go out tonight?
Response: I'd love to, but I'm beat. Let's stay in, watch a movie, and order some pizza.

Input: {user_input}
Response:
"""

# Let's see how we formulate a prompt with this template
target_input = "What are you planning to do this weekend?"
completed_prompt = few_shot_prompt_template.format(user_input=target_input)
print("--- Formulated Few-Shot Prompt ---")
print(completed_prompt)
````

## Part 2: Chain-of-Thought (CoT) Prompting

**Chain-of-Thought** is simple yet revolutionary. By explicitly asking the AI to "think step-by-step" or break down its reasoning first, we get significantly more accurate, logical, and transparent answers.

This is especially useful for a digital twin when reasoning through complex, personal decisions.

````python
# Example of a Chain-of-Thought Prompt for your Digital Twin
cot_prompt_template = """
You are my digital twin. Solve this decision-making dilemma by thinking step-by-step.

Structure your response exactly like this:
Reasoning Process:
1. [Analyze key factors]
2. [Weigh the pros and cons]
3. [Draft your ultimate choice]

Decision: [Final decision here]

Input: {user_decision}
"""

test_decision = "Should I study for another 2 hours tonight for my medical entrance exam, or sleep early to be fresh?"
completed_cot_prompt = cot_prompt_template.format(user_decision=test_decision)

print("--- Generated Chain-of-Thought Prompt ---")
print(completed_cot_prompt)
````

### 🚀 Run the Chain-of-Thought Prompt via Gemini

Run the cell below to see your digital twin reason step-by-step!

````python
from google import genai
from google.colab import userdata

try:
    # Use the official new google-genai client library
    api_key = userdata.get('GOOGLE_API_KEY')
    client = genai.Client(api_key=api_key)

    print("Sending Chain-of-Thought prompt to Gemini...")
    response = client.models.generate_content(
        model='gemini-3.8-flash',
        contents=completed_cot_prompt,
    )

    print("\n--- Twin's Reasoned Response ---")
    print(response.text)

except Exception as e:
    print(f"Could not complete request: {e}")
````
